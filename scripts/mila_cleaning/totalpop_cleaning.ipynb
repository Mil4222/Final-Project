```python
#setup
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
#paths
data_path = Path("../../data")
population_path = data_path / "total_pop.csv"
```

```python
#load data
population = pd.read_csv(population_path)
```

```python
#look before cleaning
print(population.shape)
print(population.info())
```

```python
print("\nUnique countries count:", population["REF_AREA"].nunique())
print("Years available:", sorted(population["TIME_PERIOD"].unique()))
```

```python
#useful columns kept
population_clean = population[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].copy()
```

```python
#rename for consistency
population_clean = population_clean.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "population"
})
```

```python
#convert to integer
population_clean["population"] = population_clean["population"].astype(int)
population_clean = population_clean.dropna(subset=["population"])
```

```python
#look at after cleaning
print(population_clean.shape)
print(population_clean.info())
print(population_clean["population"].describe())


print("\nMissing values in clean dataset:")
print(population_clean.isna().sum())
```
