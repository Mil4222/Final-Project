```python
import pandas as pd 
import numpy as np 
from pathlib import Path 
```

```python
data_path=Path("../../data")
gdp_cacu=data_path/"gdp_capita_current_us.csv"
```

```python
gdp=pd.read_csv(gdp_cacu, keep_default_na=False, na_values=[""])
```

```python
print(gdp.shape)
print(gdp.info())
print(gdp.head)
print(gdp.isna().sum())
```

```python
# check that file contains only one series (all should be 1)
print(gdp[["FREQ", "INDICATOR", "UNIT_MEASURE", "UNIT_MULT", "SEX", "AGE", "URBANISATION"]].nunique())
print(gdp["INDICATOR_LABEL"].unique())  # -> GDP per capita (current US$)

# if sex -> male female nonbinary then check that year NOT three times 
```

```python
# rename 
gdp_clean = gdp.rename(columns={
    "REF_AREA": "iso3",  
    "REF_AREA_LABEL": "country", 
    "TIME_PERIOD": "year",  
    "OBS_VALUE": "gdp per capita", #!UNSURE 
})
```

```python
# year integer
gdp_clean["year"] = gdp_clean["year"].astype(int)  # make sure year whole number
```

```python
# keep only useful columns 
gdp_clean = gdp_clean[["iso3", "country", "year", "gdp per capita"]]
```

```python
gdp_clean.isna().sum()  # missing values -> all 0
```

```python
# export clean data
output_dir = data_path / "clean_data_mila"  
output_dir.mkdir(parents=True, exist_ok=True)  
gdp_clean.to_csv(output_dir / "gdp_clean2.csv", index=False) 
```
