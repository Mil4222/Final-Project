```python
#set up 
import pandas as pd
import numpy as np 
from pathlib import Path 
```

```python
#paths
data_path= Path("../../data")
rural_path=data_path/"rural_pop_tot.csv"
```

```python
#import data
rural_pop= pd.read_csv(rural_path)
```

```python
print(rural_pop.info)
print(rural_pop.shape)
display(rural_pop.head(3))
```

```python
print(rural_pop.isna().sum()[rural_pop.isna().sum() > 0])
```

```python
print("\nUrbanisation categories:", rural_pop["URBANISATION_LABEL"].unique())
```

```python
#keep only useful
rural_clean = rural_pop[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].copy()
```

```python
#rename
rural_clean = rural_clean.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "rural_population"
})
```

```python
#convert to int because more sense 
rural_clean["rural_population"] = rural_clean["rural_population"].astype(int)
```

```python
#look after cleaning

print("Cleaned Dataset Shape:", rural_clean.shape)
print(rural_clean["rural_population"].describe())
```
