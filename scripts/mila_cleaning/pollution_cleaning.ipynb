```python
# Import 
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
#define path 
data_path = Path("../../data")
pollution_path = data_path / "pollution_mean_annual_exposure.csv"
```

```python
#load data set 
pollution = pd.read_csv(pollution_path)
```

```python
print(pollution.info())
print(pollution.shape)
```

```python
# summary statistics 
print(pollution.head())
```

```python
print(pollution.columns.tolist())
```

```python
print(pollution.isna().sum()[pollution.isna().sum() > 0])
```

```python
#keep useful
pollution_clean = pollution[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].copy()
```

```python
#rename
pollution_clean = pollution_clean.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "pm2.5_pollution"
})
```

```python
#rounding to two decimals 

pollution_clean["pm2.5_pollution"]=pollution_clean["pm2.5_pollution"].round(2)
```

```python
#look at data after cleanin
print(pollution_clean.shape)
print(pollution_clean.info)
print(pollution_clean["pm2.5_pollution"].describe())

```

```python
latest_year = pollution_clean["year"].max()
print(f"\nTop 5 Most Polluted Countries ({latest_year}):")
print(pollution_clean[pollution_clean["year"] == latest_year]
      .sort_values(by="pm2.5_pollution", ascending=False)[["country", "pm2.5_pollution"]].head(5))
```
