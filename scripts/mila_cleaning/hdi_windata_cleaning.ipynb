```python
import pandas as pd
import numpy as np 
from pathlib import Path 
```

```python
data_path=Path("../../data")
hdi_windatapath=data_path/"hdi_windata.csv"
```

```python
hdi1=pd.read_csv(hdi_windatapath)
```

```python
print(hdi1.shape)
print(hdi1.info())
print(hdi1.head)
print(hdi1.isna().sum())
```

```python
# rename consistent with other data sets
hdi1_clean = hdi1.rename(columns={
    "Entity": "country",
    "Code": "iso3",
    "Year": "year",
    "Human Development Index": "hdi",
    "World region according to OWID": "region"
})
```

rows not containing iso3 or hdi socre NOT dropped 

```python
#year integer 

hdi1_clean["year"]=hdi1_clean["year"].astype(int)
```

```python
#export clean data 

output_dir = data_path / "clean_data_mila"
output_dir.mkdir(parents=True, exist_ok=True)
hdi1_clean.to_csv(output_dir / "hdi_windata_clean.csv", index=False)
```
