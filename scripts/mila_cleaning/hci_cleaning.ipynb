```python
# Import 
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
data_path=Path("../../data")
hci_path=data_path/"hci.csv"
```

```python
hci=pd.read_csv(hci_path)
```

```python
print(hci.shape)
print(hci.info())
```

```python
print(hci.REF_AREA)
```

```python
#  how many unique countries/territories in dataset
print("Number of unique countries:", hci["REF_AREA"].nunique())

# View a list of unique country labels
print(hci["REF_AREA_LABEL"].unique())
```

```python
# 1. Inspect unique values in SEX 
print("Original SEX categories:", hci["SEX"].unique())
```

```python
#only keep total values leave out female and male
hci_clean = hci[hci["SEX"] == "_T"].copy()
```

```python
print(hci.AGE_LABEL) # no useful info
```

```python
print(hci.OBS_VALUE)
```

```python
#only useful kept
hci_clean = hci_clean[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]]
```

```python
# Rename columns 
hci_clean = hci_clean.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "hci_value"
})
```

```python
hci_clean = hci_clean.dropna(subset=["hci_value"]).copy()
hci_clean["hci_value"] = hci_clean["hci_value"].round(3)
print(hci_clean.shape)   # expect (601, 4)
```

```python
print(hci_clean)
```

```python
print(hci_clean["hci_value"].describe())
```

```python
output_dir = data_path / "clean_data_mila"
output_dir.mkdir(parents=True, exist_ok=True)
hci_clean.to_csv(output_dir / "hci_clean.csv", index=False)
```
