```python
urbanpop=urbanpop.drop(columns="country")
unemployment=unemployment.drop(columns="country")

rural=rural.drop(columns="country")
pollution=pollution.drop(columns="country")
hci=hci.drop(columns="country")

hdi = hdi.drop(columns={"country", "region"})
life = life.drop(columns="country")
```

```python
#w/o hci and life = 178, w/o hci = 168, with everything = 156
common_iso=(set(geo["iso3"])&set(unemployment["iso3"])&set(urbanpop["iso3"])&set(pollution["iso3"])&set(rural["iso3"])&set(hdi["iso3"])&set(life["iso3"])&set(hci["iso3"]))

# &set(hci["iso3"])
# &set(life["iso3"])
# &set(hdi["iso3"])
print(len(common_iso))
```

```python
hdi_only = set(hdi["iso3"]) - set(geo["iso3"])
hdi[hdi["iso3"].isin(hdi_only)][["iso3", "country"]].drop_duplicates()
```

```python
geo_only = set(geo["iso3"]) - set(hdi["iso3"])
geo[geo["iso3"].isin(geo_only)][["iso3", "country"]].drop_duplicates()
```

```python
geo_only = set(geo["iso3"]) - set(life["iso3"])
geo[geo["iso3"].isin(geo_only)][["iso3", "country"]].drop_duplicates()
```

```python
geo=geo[geo["iso3"].isin(common_iso)].copy()

unemployment=unemployment[unemployment["iso3"].isin(common_iso)].copy()
urbanpop=urbanpop[urbanpop["iso3"].isin(common_iso)].copy()

hci=hci[hci["iso3"].isin(common_iso)].copy()
pollution=pollution[pollution["iso3"].isin(common_iso)].copy()
rural=rural[rural["iso3"].isin(common_iso)].copy()
```

```python
final=geo.merge(unemployment,on=["iso3", "year"], how="outer")
```

```python
final=final.merge(urbanpop,on=["iso3","year"], how="outer")
```

```python
final=final.merge(hci, on=["iso3", "year"], how="outer")
```

```python
final=final.merge(rural, on=["iso3", "year"], how="outer")
```
