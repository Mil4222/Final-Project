```python
# drop column country -> only keep the geo one as base
urbanpop = urbanpop.drop(columns="country")
unemployment = unemployment.drop(columns="country")
rural = rural.drop(columns="country")
hdi = hdi.drop(columns="country")
life = life.drop(columns="country")
age = age.drop(columns="country")
gdp = gdp.drop(columns="country")
edu = edu.drop(columns="country")
```

```python
# check how many countries are in common
# w/o age: 178, with everything: 178 --> keep age 
common_iso=(set(geo["iso3"])&set(unemployment["iso3"])&set(urbanpop["iso3"])&set(rural["iso3"])&set(hdi["iso3"])&set(life["iso3"])&set(age["iso3"])&set(gdp["iso3"])&set(edu["iso3"]))

print(len(common_iso))
# Final decision: get rid of hci and keep hdi
```

```python
# for each dataset, keep only the common iso3 to prepare for merging 
geo = geo[geo["iso3"].isin(common_iso)].copy()
unemployment = unemployment[unemployment["iso3"].isin(common_iso)].copy()
urbanpop = urbanpop[urbanpop["iso3"].isin(common_iso)].copy()
rural = rural[rural["iso3"].isin(common_iso)].copy()
hdi = hdi[hdi["iso3"].isin(common_iso)].copy()
life = life[life["iso3"].isin(common_iso)].copy()
age = age[age["iso3"].isin(common_iso)].copy()
gdp = gdp[gdp["iso3"].isin(common_iso)].copy()
edu = edu[edu["iso3"].isin(common_iso)].copy()
```

Merging plan:
- keep only common countries (iso3) -> get rid of aggregates, small territories and differences
- keep all possible years available (can be useful for future analysis, but maybe not?)
- for now: merge on iso3 and year 
- outer merge (keep all countries)

```python
final = geo.merge(unemployment, on=["iso3", "year"], how="outer")
```

```python
final = final.merge(urbanpop, on=["iso3","year"], how="outer")
```

```python
final = final.merge(hdi, on=["iso3", "year"], how="outer")
```

Variables:
- iso3: 176 countries in total
- year: 1960-2025
- population: total population number (1960-2025)
- forest_ratio: forest area as a percentage of total area (1990-2024)
- pm2.5_pollution: mean annual exposure PM2.5 air pollution (1990-2023)
- country: country name (GeoNames data)
- area_sq_km: total country area in square kilometers
- continent: Asia = AS, Africa = AF, Europe = EU, South America = SA, North America = NA, Oceania = OC
- capital_latitude: latitude of the country's capital
- capital_longitude: longitude of the country's capital
- capital: capital name (GeoNames data)
- pop_density: population/area_sq_km 
- unemployment_rate: adult unemployment rate as % of total labor force (1991-2024)
- urban_pop_total: total urban population (1960-2025)
- urban_pop_perc: urban population as % of total population (1960-2025)
- urban_pop_growth: urban population growth (1961-2025)
- hdi: human developement index (0-1) (1990-2023)
- rural_population: total rural population (1960-2025)
- life_expectancy: life expectancy at birth (total female + male) (1960-204)
- median_age: median age (1960-2023)
- distance_equator: just absolute value of latitude (distance as absolute degrees of latitude) -> based on capital's latitude
- gdp_capita: GDP per capita, current US dollar (1960-2025)
- secondary_edu_years: duration in years of secondary education (1970-2023)
