```python
import pandas as pd
import numpy as np

from pathlib import Path
data_path = Path("../../data")
out_path = data_path / "clean_data_ivana"
```

```python
life = pd.read_csv(data_path / "life.csv")
```

```python
print(life.shape)
print(life.columns)
print(life.info())
print(life.isnull().sum())
```

```python
life = life.rename(columns={
    "REF_AREA_LABEL": "country",
    "REF_AREA": "iso3",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "life_expectancy"
})
```

```python
life = life[["country", "iso3", "year", "life_expectancy"]].copy()
```

```python
life.isna().sum()
```

```python
life["life_expectancy"] = life["life_expectancy"].round(3)
```

```python
life.to_csv(out_path / "life_expectancy.csv", index=False)
```
