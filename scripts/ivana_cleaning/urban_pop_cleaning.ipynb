```python
# data set 3
urb_growth = pd.read_csv(data_path / "urb_pop_growth.csv")
print(urb_growth.shape)
urb_growth.head()
```

```python
urb_growth.columns.tolist()
```

```python
urb_growth = urb_growth[["REF_AREA", "REF_AREA_LABEL", "SEX", "AGE", "URBANISATION", "TIME_PERIOD", "OBS_VALUE"]]
urb_growth.head(10)
```

```python
# check
print(urb_growth["SEX"].unique())
print(urb_growth["AGE"].unique())
print(urb_growth["URBANISATION"].unique())
```

```python
urb_growth = urb_growth[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]]
urb_growth = urb_growth.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "urban_pop_growth"
})
urb_growth.head()
```

```python
# check
print(urb_growth.isna().sum())
```

```python
# merge of all three datasets on urbanisation
urban = urb_tot.merge(urb_perc.drop(columns="country"), on=["iso3", "year"], how="outer")
urban = urban.merge(urb_growth.drop(columns="country"), on=["iso3", "year"], how="outer")
urban = urban.sort_values(["iso3", "year"]).reset_index(drop=True)
urban.head(10)
```

```python
print(urban.shape)
print(urban.duplicated(subset=["iso3", "year"]).sum())
print(urban.isna().sum())
```

```python
urban[urban["country"].isna()]
```

```python
names = dict(zip(urb_perc["iso3"], urb_perc["country"]))
urban["country"] = urban["country"].fillna(urban["iso3"].map(names))
print(urban["country"].isna().sum())
```

```python
print(urban.isna().sum())
```
