```python
import pandas as pd
import numpy as np

from pathlib import Path
data_path = Path("../../data")
out_path = data_path / "clean_data_ivana"
```

```python
hdi = pd.read_csv(data_path / "HDI_Data.csv")
```

```python
print(hdi.shape)
print(hdi.columns)
print(hdi.info())
print(hdi.isnull().sum())
```

```python
hdi.dropna(inplace=True)
print(hdi.isnull().sum())
```

```python
hdi = hdi.rename(columns={
    "Country Name": "country",
    "Country Code": "iso3",
    "Time": "year",
    "Life expectancy at birth, total (years) [SP.DYN.LE00.IN]": "life_expectancy"
})
```

```python
hdi = hdi[["country", "iso3", "year", "life_expectancy"]].copy()
```

```python
(hdi["life_expectancy"] == "..").sum()
```

```python
hdi["life_expectancy"] = pd.to_numeric(hdi["life_expectancy"], errors="coerce")
```

```python
hdi.isna().sum()
```

```python
hdi["year"] = hdi["year"].astype("int64")
```

```python
hdi.to_csv(out_path / "life_expectancy.csv", index=False)
```
