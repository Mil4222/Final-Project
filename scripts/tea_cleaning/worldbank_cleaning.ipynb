```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# relative path
data_path = Path("../../data")

population_path = data_path / "total_pop.csv"
forest_path = data_path / "forest_perc.csv"
pollution_path = data_path / "pollution_mean_annual_exposure.csv"
```

```python
# Import data
population = pd.read_csv(population_path)
forest = pd.read_csv(forest_path)
pollution = pd.read_csv(pollution_path)
```

```python
print(population.shape)
print(population.info())
```

```python
# keep only useful columns for population
population_clean = population[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].copy()
```

```python
# change column names to match geo data (population)
population_clean = population_clean.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "population"
})
```

```python
# convert population data type
population_clean["population"] = population_clean["population"].astype(int)
```

```python
# check for missing values for iso3
print(population_clean["iso3"].isna().sum())
```

```python
print(forest.shape)
print(forest.info())
```

```python
# keep only useful columns for forest
forest_clean = forest[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].copy()
```

```python
# change column names to match geo data (forest)
forest_clean = forest_clean.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "forest_ratio"
})
```

```python
# keep only two decimals in forest_ratio
forest_clean["forest_ratio"] = forest_clean["forest_ratio"].round(2)
```

```python
# check for missing values for iso3
print(forest_clean["iso3"].isna().sum())
```

```python
print(pollution.shape)
print(pollution.info())
```

```python
# keep only useful columns for pollution
pollution_clean = pollution[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].copy()
```

```python
# change column names to match geo data (pollution)
pollution_clean = pollution_clean.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "pm2.5_pollution"
})
```

```python
# keep only two decimals in pm2.5_pollution
pollution_clean["pm2.5_pollution"] = pollution_clean["pm2.5_pollution"].round(2)
```

```python
# check for missing values for iso3
print(pollution_clean["iso3"].isna().sum())
```

Prepare for the merge:
keep all years? yes.
keep all countries? -> no keep just common ones
- Forest 262
- Pollution 248
- Population 265

```python
# check for different countries
population_iso = set(population_clean["iso3"].unique())
forest_iso = set(forest_clean["iso3"].unique())
pollution_iso = set(pollution_clean["iso3"].unique())

pop_not_forest = population_iso - forest_iso
print(population_clean[population_clean["iso3"].isin(pop_not_forest)][["iso3", "country"]].drop_duplicates())

pop_not_poll = population_iso - pollution_iso
print(population_clean[population_clean["iso3"].isin(pop_not_poll)][["iso3", "country"]].drop_duplicates())

```

```python
# check how many countries are in common
common_iso = population_iso & forest_iso & pollution_iso
len(common_iso)
```

```python
# remove country column to keep final dataframe clean
forest_merge = forest_clean.drop(columns="country")
pollution_merge = pollution_clean.drop(columns="country")
```

```python
# Merge population with forest (outer to keep all years)
wb_combined = population_clean.merge(forest_merge, on=["iso3", "year"], how="outer")
```

```python
# merge also pollution
wb_combined = wb_combined.merge(pollution_merge, on=["iso3", "year"], how="outer")
```

```python
print(wb_combined.shape)
print(wb_combined.isna().sum())
wb_combined.duplicated(subset=["iso3", "year"]).sum()
```

```python
# keep only common iso3 between all datasets (common country/territory)
wb_combined = wb_combined[wb_combined["iso3"].isin(common_iso)].copy()
```

```python
# Export clean data
output_path = data_path / "clean_data_tea" / "wb_combined.csv"
wb_combined.to_csv(output_path, index=False)
```
