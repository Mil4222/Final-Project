```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
data_path = Path("../../data")
gdp_path = data_path / "constant_gdp_capita.csv"
gdp = pd.read_csv(gdp_path)
```

```python
print(gdp.shape)
print(gdp.columns)
print(gdp.dtypes)
print(gdp.head)
print(gdp.isna().sum())
```

```python
gdp = gdp[["REF_AREA", "REF_AREA_LABEL", "TIME_PERIOD", "OBS_VALUE"]].copy()
```

```python
gdp = gdp.rename(columns={
    "REF_AREA": "iso3",
    "REF_AREA_LABEL": "country",
    "TIME_PERIOD": "year",
    "OBS_VALUE": "constant_gdp_capita"
})
```

```python
print(gdp.isna().sum())
```

```python
gdp["constant_gdp_capita"] = gdp["constant_gdp_capita"].round(3)
```

```python
output_path = data_path / "clean_data_tea" / "constant_gdp_capita.csv"
gdp.to_csv(output_path, index=False)
```
