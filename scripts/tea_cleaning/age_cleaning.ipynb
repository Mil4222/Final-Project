```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
data_path = Path("../../data")
age_path = data_path / "median_age.csv"
age = pd.read_csv(age_path)
```

```python
print(age.shape)
print(age.columns)
print(age.dtypes)
print(age.head)
print(age.isna().sum())
```

```python
age = age.rename(columns={
    "Entity": "country",
    "Code": "iso3",
    "Year": "year",
    "Median age": "median_age"
})
```

```python
age = age[age["year"].between(1960, 2023)].copy()
```

```python
age = age.drop("Median age (Projected)", axis=1)
```

```python
age.isna().sum()
```

```python
age[age["iso3"].isna()][["iso3", "country"]].drop_duplicates()
```

```python
age = age.dropna()
```

```python
output_path = data_path / "clean_data_tea" / "median_age_clean.csv"
age.to_csv(output_path, index=False)
```
