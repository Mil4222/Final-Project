```python
import pandas as pd
import numpy as np
import os
```

```python
# find relative paths
os.getcwd()

from pathlib import Path

data_path = Path("../../data")

country_path = data_path / "countryInfo.txt"
cities_path = data_path / "cities15000.txt"
```

```python
# Import data
countries = pd.read_csv(country_path, sep="\t", comment="#", header=None, names=range(19), engine="python", keep_default_na=False)
cities = pd.read_csv(cities_path, sep="\t", header=None, low_memory=False, keep_default_na=False)
```

```python
print(cities.shape)
print(cities.info())
```

```python
# name columns (countries)
countries.columns = [
    "iso",
    "iso3",
    "iso_numeric",
    "fips",
    "country",
    "capital",
    "area_sq_km",
    "population",
    "continent",
    "tld",
    "currency_code",
    "currency_name",
    "phone",
    "postal_code_format",
    "postal_code_regex",
    "languages",
    "geonameid",
    "neighbours",
    "equivalent_fips_code"
]
```

```python
# name columns (cities)
cities.columns = [
    "geonameid",         
    "name",              
    "asciiname",         
    "alternatenames",    
    "latitude",          
    "longitude",         
    "feature_class",     
    "feature_code",      
    "country_code",      
    "cc2",               
    "admin1_code",       
    "admin2_code",        
    "admin3_code",       
    "admin4_code",       
    "population",         
    "elevation",         
    "dem",               
    "timezone",          
    "modification date"
]
```

```python
# keep only useful columns (countries)
countries_clean = countries[["iso", "iso3", "country", "capital", "area_sq_km", "continent"]].copy()
```

```python
# kepp only useful columns (cities)
cities_clean = cities[["geonameid", "name", "latitude", "longitude", "feature_code", "country_code"]].copy()
```

```python
# check for missing values and duplicates (iso3 -> for later merging)
print(countries_clean["iso3"].isna().sum())
print(countries_clean["iso3"].duplicated().sum())
```

```python
# create new dataframe with just capitals and rename columns
capitals = cities_clean[cities_clean["feature_code"] == "PPLC"].copy()

capitals = capitals.rename(columns={
    "name": "capital",
    "latitude": "capital_latitude",
    "longitude": "capital_longitude"
})
```

```python
# keep only useful columns (capitals)
capitals_clean = capitals[["country_code", "capital", "capital_latitude", "capital_longitude"]]
```

```python
print(capitals_clean.shape)
print(capitals_clean["country_code"].isna().sum())
print(capitals_clean["country_code"].duplicated().sum())
```
