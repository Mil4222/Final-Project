```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
data_path = Path("../../data")
edu_path = data_path / "mean_year_school.csv"
edu = pd.read_csv(edu_path)
```

```python
print(edu.shape)
print(edu.columns)
print(edu.dtypes)
print(edu.head)
print(edu.isna().sum())
```

```python
edu = edu.rename(columns={
    "Entity": "country",
    "Code": "iso3",
    "Year": "year",
    "Both genders": "mean_edu_year"
})
```

```python
edu.isna().sum()
```

```python
edu[edu["iso3"].isna()][["iso3", "country"]].drop_duplicates()
```

```python
edu = edu.dropna()
```

```python
output_path = data_path / "clean_data_tea" / "mean_year_school_clean.csv"
edu.to_csv(output_path, index=False)
```
